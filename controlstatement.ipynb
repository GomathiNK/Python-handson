```python
Total=int(input("Enter the total mark"))
if(Total<35):
    print("Fail")
elif(Total<=65):
    print("Third class")
elif(Total<=75):
    print("Second class")
elif(Total<=85):
    print("first class")
else:
    print("Distinction")
```
