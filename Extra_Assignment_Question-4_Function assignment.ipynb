```python
def OddEven():
    num=int(input("Enter a number"))
    if((num%2)==0):
        print(num,"is Even number")
        temp="Even number"
    else:
        print(num,"is Odd number")
        temp="Even number"
    return temp
```

```python
def percentage(Sub1,Sub2,Sub3,Sub4,Sub5):
    print("Subject1=",Sub1)
    print("Subject2=",Sub2)
    print("Subject3=",Sub3)
    print("Subject4=",Sub4)
    print("Subject5=",Sub5)
    Tot=Sub1+Sub2+Sub3+Sub4+Sub5
    Per=Tot/5
    print("Total:",Tot)
    temp="Tot"
    print("Percentage:",Per)
    temp="Percentage"
    return temp
percentage(98,87,95,95,93)
```
